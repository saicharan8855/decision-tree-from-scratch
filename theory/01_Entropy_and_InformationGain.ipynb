# Entropy and Information Gain

## Introduction

A decision tree learns by asking questions.

Example:

- "Is age > 30?"
- "Did the user click before?"
- "Is salary high?"

At every node, the tree must choose:

> Which feature gives the best split?

A good split should separate classes clearly.

Example:

| Group | Result |
|---|---|
| Mostly Spam | Good split |
| Mostly Not Spam | Good split |

Bad split:

| Group | Result |
|---|---|
| Mixed classes everywhere | Bad split |

So we need a way to measure:

> How mixed is the data?

This mixedness is called **impurity**.

- Pure node $\rightarrow$ only one class
- Impure node $\rightarrow$ many classes mixed

Entropy measures this impurity.

---

## The Cauchy Axioms

Suppose uncertainty is measured by:

$$
H(p_1,p_2,\dots,p_n)
$$

where

$$
p_i = \text{probability of class } i
$$

The measure should satisfy some natural rules.

---

### 1. Continuity

If probabilities change slightly,

then uncertainty should also change slightly.

Small change in input  
$\downarrow$

Small change in output



### 2. Maximum for Uniform Distribution

If all outcomes are equally likely,

uncertainty should be maximum.

Example:

$$
(0.5,0.5)
$$

is more uncertain than

$$
(0.99,0.01)
$$

because both outcomes are equally possible.



### 3. Additivity

If a problem is solved in multiple stages,

total uncertainty should equal:

$$
\text{uncertainty of first stage}
+
\text{expected uncertainty of later stages}
$$

This means uncertainty should combine consistently.



From these axioms, Shannon proved:

$$
H = -K \sum p_i \log p_i
$$

For information theory:

$$
K = 1
$$

Using base 2 logarithm:

$$
H = -\sum p_i \log_2 p_i
$$

---

## Entropy Derivation

Start from the additivity axiom.

Suppose all outcomes are equally likely.

Let

$$
A(n)
$$

be uncertainty for $n$ equally likely outcomes.



If two independent choices happen:

- first choice has $m$ outcomes
- second choice has $n$ outcomes

then total outcomes become:

$$
mn
$$

Additivity says:

$$
A(mn)=A(m)+A(n)
$$

We need a function satisfying:

$$
f(mn)=f(m)+f(n)
$$

The logarithm satisfies this property:

$$
\log(mn)=\log m+\log n
$$

Therefore:

$$
A(n)=K\log n
$$



Now move to unequal probabilities.

Suppose:

$$
p_i=\frac{n_i}{N}
$$

where

$$
\sum n_i = N
$$

Choose one item from $N$ items.

Total uncertainty:

$$
K\log N
$$

But the selection happens in two stages:

1. Choose group
2. Choose item inside group

Uncertainty inside group $i$:

$$
K\log n_i
$$

Expected uncertainty:

$$
\sum p_i K\log n_i
$$

Using additivity:

$$
K\log N
=
H(p_1,\dots,p_n)
+
\sum p_i K\log n_i
$$

Move terms:

$$
H
=
K\log N
-
K\sum p_i \log n_i
$$

Factor $K$:

$$
H
=
K\left(
\log N
-
\sum p_i \log n_i
\right)
$$

Since:

$$
p_i=\frac{n_i}{N}
$$

then:

$$
n_i=p_iN
$$

Substitute:

$$
H
=
K\left(
\log N
-
\sum p_i \log(p_iN)
\right)
$$

Expand logarithm:

$$
\log(p_iN)=\log p_i+\log N
$$

So:

$$
H
=
K\left(
\log N
-
\sum p_i(\log p_i+\log N)
\right)
$$

Distribute summation:

$$
H
=
K\left(
\log N
-
\sum p_i\log p_i
-
\sum p_i\log N
\right)
$$

Since:

$$
\sum p_i = 1
$$

then:

$$
\sum p_i\log N = \log N
$$

Thus:

$$
H
=
K\left(
\log N
-
\sum p_i\log p_i
-
\log N
\right)
$$

Therefore:

$$
H=-K\sum p_i\log p_i
$$

Using base 2 logarithm and $K=1$:

$$
H=-\sum p_i\log_2 p_i
$$

---

## Properties of Entropy

For binary classification:

$$
H(p)=-p\log_2 p-(1-p)\log_2(1-p)
$$

If one class dominates:

$$
p \rightarrow 1
$$

then:

$$
H \rightarrow 0
$$

because uncertainty disappears.



Example:

$$
p=1
$$

$$
H=0
$$

Perfect purity.



Entropy is maximum when classes are equally likely.

For binary case:

$$
p=0.5
$$

$$
H(0.5)=1
$$

Maximum uncertainty.



Shape of graph:

- near $0$ $\rightarrow$ low entropy
- near $1$ $\rightarrow$ low entropy
- middle $\rightarrow$ high entropy

---

## Conditional Entropy

Suppose:

- $Y$ = target class
- $X$ = feature

Conditional entropy:

$$
H(Y|X)
$$

means:

> remaining uncertainty in $Y$ after knowing $X$



Formula:

$$
H(Y|X)
=
\sum_x P(x)H(Y|X=x)
$$

Each split creates subsets.

Entropy is computed inside every subset.

Then weighted average is taken.



Why weighted?

Because larger subsets should influence more.

Example:

| Subset Size | Importance |
|---|---|
| 90 samples | High |
| 10 samples | Low |

So probabilities become weights.

# Information Gain

Information Gain measures:

> how much uncertainty is reduced after split

Formula:

$$
IG(Y,X)=H(Y)-H(Y|X)
$$

- before split $\rightarrow H(Y)$
- after split $\rightarrow H(Y|X)$

Difference $=$ gained information.



Suppose dataset:

| Class | Count |
|---|---|
| Yes | 6 |
| No | 4 |

Total:

$$
10
$$



Parent entropy:

$$
H(Y)
=
-\frac{6}{10}\log_2\frac{6}{10}
-
\frac{4}{10}\log_2\frac{4}{10}
$$

$$
H(Y)\approx0.971
$$



Now split by feature.

Left child:

| Class | Count |
|---|---|
| Yes | 4 |
| No | 1 |

Entropy:

$$
H_L
=
-\frac45\log_2\frac45
-
\frac15\log_2\frac15
$$

$$
H_L\approx0.722
$$



Right child:

| Class | Count |
|---|---|
| Yes | 2 |
| No | 3 |

Entropy:

$$
H_R
=
-\frac25\log_2\frac25
-
\frac35\log_2\frac35
$$

$$
H_R\approx0.971
$$



Weighted entropy:

$$
H(Y|X)
=
\frac{5}{10}(0.722)
+
\frac{5}{10}(0.971)
$$

$$
H(Y|X)\approx0.846
$$



Information gain:

$$
IG
=
0.971-0.846
$$

$$
IG\approx0.125
$$

Higher information gain  
$\rightarrow$ better split.

## Why log base 2

Different logarithm bases give different units.



Base 2:

$$
\log_2
$$

Unit:

$$
\text{bits}
$$



Base $e$:

$$
\ln
$$

Unit:

$$
\text{nats}
$$



Base 10:

$$
\log_{10}
$$

Unit:

$$
\text{hartleys}
$$



Changing base only multiplies entropy by a constant.

Using change of base formula:

$$
\log_b x
=
\frac{\log_k x}{\log_k b}
$$

So rankings do not change.

If one split has higher entropy using base 2,

it will also have higher entropy using base $e$.

Thus split selection remains same.

## Limitations

Entropy prefers features with many unique values.

Example:

| Student ID | Class |
|---|---|
| 101 | Yes |
| 102 | No |
| 103 | Yes |

Every split becomes perfectly pure.

Entropy becomes very low.

Information gain becomes very high.

But this split is useless because IDs do not generalize.



This is called:

> High Cardinality Problem



Fix:

$$
\text{Gain Ratio}
=
\frac{\text{Information Gain}}{\text{Split Information}}
$$

Gain Ratio penalizes overly large splits.

Used in:

- C4.5 Decision Trees
