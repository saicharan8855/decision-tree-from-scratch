```python
import numpy as np
import sys
sys.path.append('src')
from decision_trees import DecisionTree
```

```python
# Load breast cancer processed arrays
X_train = np.load('../Data/processed/bc_X_train.npy')
y_train = np.load('../Data/processed/bc_y_train.npy')
X_test  = np.load('../Data/processed/bc_X_test.npy')
y_test  = np.load('../Data/processed/bc_y_test.npy')

print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
print("X_test  shape:", X_test.shape)
print("y_test  shape:", y_test.shape)
```

```python

tree = DecisionTree(
    criterion="gini",
    max_depth=3,
    task="classification"
)

tree.fit(X_train, y_train)
print("Tree trained successfully.")
```

```python

y_pred = tree.predict(X_test)


print(f"{'Index':<8} {'Predicted':<12} {'Actual':<10} {'Match'}")
print("-" * 40)

for i in range(10):
    match = "yes" if y_pred[i] == y_test[i] else "no"
    print(f"{i:<8} {y_pred[i]:<12} {y_test[i]:<10} {match}")
```

```python

correct = np.sum(y_pred == y_test)
total = len(y_test)
accuracy = correct / total

print(f"Correct predictions : {correct} / {total}")
print(f"Accuracy            : {accuracy * 100:.2f}%") 
```
