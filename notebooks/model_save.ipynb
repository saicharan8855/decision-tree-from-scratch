```python
import os
import pickle
import numpy as np
import sys

sys.path.append(os.path.join(os.getcwd(),'..','src'))


from decision_trees import DecisionTree

X_train = np.load('../Data/processed/bc_X_train.npy')
y_train = np.load('../Data/processed/bc_y_train.npy')
X_test = np.load('../Data/processed/bc_X_test.npy')
y_test = np.load('../Data/processed/bc_y_test.npy')


model = DecisionTree(
    criterion='gini',
    max_depth=5,
    task='classification'
)
model.fit(X_train, y_train)


os.makedirs('results/models', exist_ok=True)
with open('results/models/best_model.pkl', 'wb') as f:
    pickle.dump(model, f)

print('Saved to Results/models/best_model.pkl')
```
