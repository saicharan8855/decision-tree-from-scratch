```python
fig , ax = plt.subplots(figsize = (8 , 6)) 
sns.countplot(x = 'diagnosis' , data = breast_cancer )
ax.grid(visible = True)
plt.title("Diagnosis Countplot")
plt.show()
```

```python
fig , ax = plt.subplots(figsize = (8 , 6)) 
sns.countplot(x = 'target' , data = heart_disease )
ax.grid(visible = True)
plt.title("Target Countplot")
plt.show()
```

```python
breast_cancer_continuous = breast_cancer.select_dtypes(include = ['float64']).columns
heart_disease_continuous = heart_disease.select_dtypes(include = ['float64' , 'int64']).columns
```

```python
breast_cancer[breast_cancer_continuous].hist(
    figsize = (20,20),
    bins = 30
)
plt.tight_layout()
plt.show()
```

```python
heart_disease[heart_disease_continuous].hist(
    figsize = (20 , 20),
    bins = 13 
)
plt.tight_layout()
plt.show()
```

```python
fig , axes = plt.subplots(1 , 3 , figsize = (18 , 5))
sns.countplot(
    x = 'cp',
    data = heart_disease,
    ax = axes[0]
)
axes[0].set_title("chest pain value counts")

sns.countplot(
    x = 'thal',
    data = heart_disease,
    ax = axes[1]
)
axes[1].set_title("Thal") 

sns.countplot(
    x = 'restecg',
    data = heart_disease,
    ax = axes[2]
)
axes[2].set_title("restecg")

plt.tight_layout()
plt.show()
```

```python
numeric_breast_cancer = breast_cancer.select_dtypes(include = ['float64'])
correlation = numeric_breast_cancer.corr()
```

```python
plt.figure(figsize = (20,15))
sns.heatmap(
    correlation,
    annot = True,
    fmt = '.2f', 
    cmap = 'coolwarm'
)
plt.title("correlation matrix")
plt.show()
```

```python
numeric_heart_disease = heart_disease.select_dtypes(include = ['float64','int64'])
correlation_hd = numeric_heart_disease.corr()
```

```python
plt.figure(figsize = (20,15))
sns.heatmap(
    correlation_hd,
    annot= True,
    fmt = '.2f', 
    cmap = 'coolwarm'
)
plt.title("Correlation matrix for heart_disease")
plt.show()
```
